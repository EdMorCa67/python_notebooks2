## OBJETOS MUTABLES

### Listas

```python

L = [2, 3, 4]	    # Una lista de tres elementos
M = L			    # Hace una referencia a L
L[0] = 24 		    # cambia el primer elemento de L

print("L = ", L)    
print("M = ", M)    # En estos dos casos se muestra que M también ha cambiado pues es una referencia de L
```

Copia de objetos en vez de referencia:


```python

L = [2, 3, 4]
M = L[:]		    # Hace una copia de la lista L en M, por lo que M no es una referencia de esta
L[0] = 24

print("L = ", L)    # L cambia

print("M = ", M)    # más en este caso M sigue igual pues es la referencia de la copia
```

### Shared References and Equality
- Operador `==` igualdad de valores.
- Operador `is` identidad de objetos.

```python

L = [1, 2, 3]
M = L 			    # como antes, M es una referencia a L

L == M 			    # el == sirve para comparar el contenido de los objetos y ver si son iguales, como en este caso
L is M 			    # Lo mismo que el anterior, pero con el comando is se comprueba si ambos objetos son el mismo
```

```python
L = [1, 2, 3]
M = [1, 2, 3]	    # M y L son diferentes objetos pero poseen el mismo contenido

print(
    "L == M",
    L == M 			# Es verdad pues poseen el mismo contenido
)

print(
    "L is M", 	    # Sale falso pues son diferentes objetos
    L is M
)
```

#### CACHE

Because small integers and strings are cached and reused, though, is tells us they reference the same single object.


```python

X = 42
Y = 42		    # Should be two different objects

X  == Y         # True
X is Y 		    # True: Same object anyhow: caching at work!!
```

Averiguar el numero de refencias a un objeto:


```python
import sys
sys.getrefcount(1)		# number of pointers to this shared piece of memory
```
