## GLOBAL SCOPE

- Global names are variables assigned at the top level of the enclosing module file.
- Global names must be declared only if they are assigned within a function.
- Global names may be referenced within a function without being declared.

```python
x = 88 		# variable global x

def func():
	x = 99 	# en la función se crea una variable local x que no afecta a la variable global x

func()
print(x) #	 x se printea igual a su variable global
```

```python
x = 88 			# Global X

def func():
	global x
	x = 99 		# en este caso se modifica la variable global x

func()
print("x =", x) # ahora si sale 99
```

## LEGB rule

```python


y, z = 1, 2         # variables globales z , y

def all_global():
	global x        # Declaramos a x como variable global dentro de la función
	x = y + z       # No hace falta declarar y o z
                    # x existe ahora en el ambito global con valor 3

all_global()
print('x = %d, y = %d, z = %d' % (x, y, z))
```

## NESTED SCOPES


```python

x = 99                      # variable global

def f1():
    x = 88                  # variable local de x
    def f2():               # función dentro de función
        print("x local = ", x)    # Referencia dentro de la función anidada
    f2()                    # f2 es una función temporal                    

f1()                        # printea 88 al estar dentro de la función local
print("x global = ", x)     # salida: X = 88; X sigue valiendo 99

# f2()  NameError: name 'f2' is not defined  No se puede invocar a f2() desde el modulo principal
```

## Factory Functions: Closures

A **closure** or a **factory function**, the former describing a **functional programming technique**, and the latter denoting a **design pattern**.
The function object in question remembers values in enclosing scopes regardless of whether those scopes are still present in memory.

```python
def f1():
	x = 88                 # variable local
	def f2():
		print("x =", x)    # en la función anidada se printea la local
	return f2 	           # returna f2 pero no la llama                        

action = f1()       # action es ahora una función
action() 			# ahora lo llama
```

```python
def maker(n):
	def action(x):		# crea y retorna acción
		return x ** n 	# la acción retiene n de la función
	return action

f = maker(2)            # retorna x ** 2
f(3)                    # pasa 3 a x

f(4)		            # pasa 4 a x, retorna 16
```
