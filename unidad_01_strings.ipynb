## 2- Sequence operations

As sequences, strings support operations that assume a positional ordering among
items.

```python
S = 'Spam'
len(S) #Give the length of the string S
```

### The first item in S, indexing by zero-based position
In Python, indexes are coded as offsets from the front, and so start from `0:` the first item is at index `0`, the second is at index `1`, and so on.


```python
S[0] #first example , it gives the first character of the string
```

```python
S[1] #second example, giving the second character of the string
```

```python
S[-1] #Last example, being negative it gives the last character of the string
```

```python
S[len(S)-1] #negative index being way too complicated , its better the former way
```

```python
S[1:3]  #A range of characters from the string, starting at index 1 and ending at index 3
```

```python
S[1:] #Everything past the first letter
```

```python
# Everything but the last
S[0:3]
```

```python
# Same as S[0:3]
S[:3]
```

```python
# Everything but the last again, but simpler (0:-1)
S[:-1]
```

```python
# All of S as a top-level copy (0:len(S))
S[:]
```

Finally, as sequences, strings also support concatenation with the plus sign (joining two
strings into a new string) and repetition (making a new string by repeating another)

```python
S * 7 #repetition of strings
```

```python
S + "pancreatico" #example ofconcatenation of strings
```

## 3 - Immutability

Every string operation is defined to produce a new
string as its result, because strings are immutable in Python—they cannot be changed
in place after they are created. You can never overwrite the values of
immutable objects.

```python
S[0] = 'z' #Due to strings being inmutable, this will give an error
```

```python
S = 'z' + S[1:] #but we can run this to change the first letter of the string S
S
```

## 5 - Type-Specific Methods
strings have operations all
their own, available as methods—functions that are attached to and act upon a specific
object, which are triggered with a call expression.

```python

S = 'Spam'
S.find('pa')  #finds the first occurrence of 'pa' in S
```

```python
S.find('m', 3) #returns the index of the first occurrence of 'm' in S starting from index 3
```

```python
S.find(S) #returns 0 because the string is found at the beginning of itself
```

```python

S.find('z') #if the character doesnt exist it returns 0
```

### Replace

### Split

```python
S = 'aaa,bbb,ccccc,dd'
S.split(',') #splits the string into a list of strings using the comma as a separator 
```

### Rstrip
Remove whitespace characters on the right side

```python
line = 'aaa,bbb,ccccc,dd\n'
print(line)
```

```python
line.rstrip() #removes the trailing whitespace characters
```

```python
line.lstrip() #removes the leading whitespace characters
```

```python
# Combine two operations
line.rstrip().split(',')
```

### Formatting
Strings also support an advanced substitution operation known as **formatting**

```python
# # Formatting expression
'%s, eggs, and %s' % ('spam', 'SPAM!')
```

```python
# Formatting method
'{}, eggs, and {}'.format('spam', 'SPAM!')
```

```python
S = 'SpamEggsSpam'
f'{S}'
```

```python
'{0}'.format(S)
```

### Character code conversions

Convert a single character to its underlying integer code (e.g., its **ASCII byte value**) by passing it to the built-in `ord ` function—this returns the actual binary value used to represent the corresponding character in memory.

For ASCII text, this is the familiar 7-bit integer that fits in a single byte in memory, but the range of code points for other kinds of **Unicode** text may be wider.

```python
ord('s')
```

The `chr` function performs the inverse operation, taking an integer code and converting it to the corresponding character:

```python
chr(115)
```
