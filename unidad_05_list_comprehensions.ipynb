# Cap 14: Iterations and Comprehensions 

Cada vez que comenzamos a pensar en realizar una operación sobre cada item de una secuencia, estamos en el terreno de las _list comprehensions_.

```python
L = [1, 2, 3, 4, 5] #se hace una lista
for i in range(len(L)): #con el i in range se va por cada elemento de la lista al hacerlo con len(L) pues el codigo len devuelve el numero total de caracteres en la lista
    L[i] += 10   #a cada elemento se le suma 10
L
```

```python
L = [x + 10 for x in L] #lo mismo de antes pero abreviado en una linea
L
# [21, 22, 23, 24, 25]
```

```python
# Es lo mismo que:
res = []
for x in L:
	res.append(x + 10)
res
# [31, 32, 33, 34, 35]
```

## Extended List Comprehension Syntax

### Filter clauses: `if`

Un `for` loop anidado en una comprehension puede tener asociada una cláusula `if` para filtrar del resultado los items para los cuáles el test no es `true`.

```python

L = [1, 7, 8, 9]   #inserta una lista
M = [ x  for x in L if x < 5 ]  #M solo dará los elementos de la lista L si el valor es menor que 5
M
# [1]
```

### Bucles anidados: `for`

```python
[ x + y for x in 'abc' for y in 'lmn' ] #el x se reemplaza por cada letra en abc y se le suma cada elemento de lmn
# ['al', 'am', 'an', 'bl', 'bm', 'bn', 'cl', 'cm', 'cn']
```

```python
# Es lo mismo que:
res = []
for x in 'abc':
	for y in 'lmn':
		res.append(x + y)
res
# ['al', 'am', 'an', 'bl', 'bm', 'bn', 'cl', 'cm', 'cn']∫
```

### Tuples (pag 279)

Podemos usar las _List comprehensions_ para convertir tuples. 

```python
# Crea una lista a partir de una tupla, añadiendo 20 a cada item:
T = (1, 2, 3, 4, 5)
L = [ x + 20 for x in T ]
L
# [21, 22, 23, 24, 25]
```

### Range

```python
#inserta en la tupla los numeros de 0 a 9
L = [ x for x in range(0,9) ]
L
# [0, 1, 2, 3, 4, 5, 6, 7, 8]
```

## Dictionary comprehensions in 3.X and 2.7

### Ejemplo 1

#### zip(*iterables)

Crear un iterator que agrega elementos de cada uno de los iterables.

[funcion zip en librería Python](https://docs.python.org/3.3/library/functions.html#zip)

```python
x = [1, 2, 3]
y = [4, 5, 6]
zipped = zip(x, y) #con el comando zip se juntan ambos iterables de cada posicion con el de la misma posicion de la otra
list(zipped)
# [(1, 4), (2, 5), (3, 6)]
```

Devuelve un *iterator* de tuplas, donde la i-ésima tupla contiene el i-ésimo elemento de cada una de las secuencias o *iterables* proporcionados como argumentos. 

El *iterator* se detiene cuando el iterable de entrada más corto se agota.

```python
L = [ x for x in range(0,4) ]  #hace un iterable de 0 a 4
M = [ x for x in range(5,10) ] #lo mismo de 5 a 10
[elemento for elemento in zip(L,M) ] #insertamos el zip de antes 
# [(0, 5), (1, 6), (2, 7), (3, 8), (4, 9)]
```

```python
# hace una lista en zip de los numeros y caracteres

list(zip(['a', 'b', 'c'], [1, 2, 3]))		
# [('a', 1), ('b', 2), ('c', 3)] 

# haz un dict con el resultado

D = dict(zip(['a', 'b', 'c'], [1, 2, 3]))	
D
# {'b': 2, 'c': 3, 'a': 1}

# Es lo mismo que:
D = {k: v for (k, v) in zip(['a', 'b', 'c'], [1, 2, 3])}
D
# {'b': 2, 'c': 3, 'a': 1}
```

### Ejemplo 2

```python
# Inicializa dict con el fromkeys

D = dict.fromkeys(['a', 'b', 'c'], 0)		
D
# {'b': 0, 'c': 0, 'a': 0} 

# lo mismo pero con una comprensión

D = { k:0 for k in ['a', 'b', 'c'] }			
D
# {'b': 0, 'c': 0, 'a': 0}
```

```python
# Otros iterables con un valor establecido

D = dict.fromkeys('spam')					
D
# {'s': None, 'p': None, 'a': None, 'm': None} 

D = { k: None for k in 'spam' }
D
# {'s': None, 'p': None, 'a': None, 'm': None}
```
